```python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
# URL of the public Iris CSV dataset

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

# Define column names because the original CSV has no header

columns = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width",
    "species"
]

# Read CSV file using Pandas

df = pd.read_csv(url, names=columns)

print("CSV file imported successfully!")
```

```python
# Display first five rows

df.head()
```

```python
# Display dataset dimensions

print("Dataset Shape:", df.shape)

# Display column information

df.info()

# Display statistical summary

df.describe()
```

```python
# Check missing values in each column

print(df.isnull().sum())
```

```python
# Calculate average petal length

average = np.mean(df["petal_length"])

print("Average Petal Length:", average)

# Find maximum petal length

maximum = np.max(df["petal_length"])

print("Maximum Petal Length:", maximum)
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="species"
)

plt.title("Number of Flowers in Each Species")
plt.xlabel("Flower Species")
plt.ylabel("Number of Flowers")

plt.xticks(rotation=15)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="species",
    y="petal_length"
)

plt.title("Petal Length Comparison")
plt.xlabel("Flower Species")
plt.ylabel("Petal Length (cm)")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="petal_length",
    y="petal_width",
    hue="species",
    palette={
        "Iris-setosa": "red",
        "Iris-versicolor": "blue",
        "Iris-virginica": "green"
    },
    s=100
)

plt.title("Petal Length vs Petal Width by Iris Species",
          fontsize=15)

plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")

plt.legend(title="Flower Species")
plt.grid(True, linestyle="--", alpha=0.4)

plt.show()
```
